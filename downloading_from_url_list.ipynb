# Downloading tiles

```python
import requests
import os
from tqdm import tqdm
import csv
```

```python
def download_file(url, dest):
    if requests.get(url).status_code != 200:
        return -1

    pc = requests.get(url).content
    file_src = os.path.join(dest, url.split('/')[-1])
    with open(file_src, 'wb') as handler:
        handler.write(pc)
    return file_src
```

```python
csv_src = r"D:\Terranum_SD\99_Data\PC_movement_tracking\Doldenhorn\Doldenhorn_DEM.csv"
dest = 'default'
dest = os.path.join(os.path.dirname(csv_src), 'tiles') if dest == 'default' else dest
os.makedirs(dest, exist_ok=True)
with open(csv_src) as f:
    list_urls = [x for row in list(csv.reader(f)) for x in row]
# list_urls = pd.read_csv(csv_src)
# list_urls = list_urls[list_urls.columns[0]].to_list()
for _, url in tqdm(enumerate(list_urls), total=len(list_urls), desc="Downloading"):
    return_code = download_file(url, dest)
    if return_code == -1:
        print("Problem with: ", url, "\ncode: ", requests.get(url).status_code)
```

```python
import random
from math import sin, cos, pi
import numpy as np

def random_point_on_circle(xy, radius=100):
    angle = np.random.uniform(0, 2 * np.pi)
    xy_prime = (
        xy[0] + radius * np.cos(angle),
        xy[1] + radius * np.sin(angle)
    )
    return xy_prime

x_mean = 1
y_mean = 2
# alpha= pi
# R = np.array([[cos(alpha), sin(alpha)],[-sin(alpha), cos(alpha)]])
xy = np.array([x_mean, y_mean])
xy_prime = random_point_on_circle(xy, 100)
# print(alpha)
print(xy, xy.shape)
print(xy_prime)
x_prime, y_prime = list(xy_prime)
print(x_prime, y_prime)
print(np.linalg.norm(xy_prime - np.array(xy)))
```
