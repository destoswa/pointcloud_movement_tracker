# Downloading tiles

```python
import requests
import os
from tqdm import tqdm
import pandas as pd
```

```python
def download_file(url, dest):
    if requests.get(url).status_code != 200:
        return -1

    pc = requests.get(url).content
    file_src = os.path.join(dest, url.split('/')[-1])
    with open(file_src, 'wb') as handler:
        handler.write(pc)
    return file_src
```

```python
dest = r"D:\Terranum_SD\2_Projects\pc_movement_tracking\vaud_fribourg_liens\test"
csv_src = r"D:\Terranum_SD\2_Projects\pc_movement_tracking\vaud_fribourg_liens\test.csv"
list_urls = pd.read_csv(csv_src)
list_urls = list_urls[list_urls.columns[0]].to_list()
for _, url in tqdm(enumerate(list_urls), total=len(list_urls), desc="Downloading"):
    return_code = download_file(url, dest)
    if return_code == -1:
        print("Problem with: ", url, "\ncode: ", requests.get(url).status_code)
```

```python
class book():
    def __init__(self, list_links):
        self.links = list_links
        self.locations = [x.loc for x in list_links]
        self.years = [x.year for x in list_links]
        self.urls = [x.url for x in list_links]


class link():
    def __init__(self, url, year, location):
        self.url = url
        self.year = year
        self.loc = location

```

```python
csv_vaud = r"E:\pointclouds_vaud_fribourg\vaud\vaud_15.csv"
# csv_tot = r"E:\pointclouds_vaud_fribourg\vaud_15.csv"

df_vaud = pd.read_csv(csv_vaud, sep=';')
list_urls = df_vaud[df_vaud.columns[0]].to_list()
print(len(set(list_urls)))
print(['_'.join(x.split('_')[1:3]).split('/')[0].replace('-','_') for x in list_urls])
print(len(['_'.join(x.split('_')[1:3]).split('/')[0].replace('-','_') for x in list_urls]))
print(len(set([x.split('_')[2].split('/')[0].replace('-','_') for x in list_urls])))
```

```python
lst_links = []
for _, x in tqdm(enumerate(list_urls), total=len(list_urls)):
    loc = x.split('_')[2].split('/')[0].replace('-', '_')
    year = x.split('_')[1]
    lst_links.append(link(x, year, loc))
mybook = book(lst_links)
```

```python
import collections
lst_duplicates = [item for item, count in collections.Counter(mybook.locations).items() if count > 1]
print(lst_duplicates)
print(set(lst_duplicates))
# lst_duplicates_indexes = [i,_ for i in mybook.]
```

```python
lst_ids_to_remove = []
for loc in lst_duplicates:
    ids = [i for i,x in enumerate(mybook.locations) if x.replace('-', '_') == loc]
    years = [x for i, x in enumerate(mybook.years) if i in ids]
    # print(ids)
    # print(years)
    # print(ids[years.index(max(years))])
    # print('--')
    id_to_rm = ids[years.index(max(years))]
    lst_ids_to_remove.append((mybook.locations[id_to_rm], mybook.years[id_to_rm]))
```

```python
src_originals = r"E:\pointclouds_vaud_fribourg\2019"
src_duplicates = r"E:\pointclouds_vaud_fribourg\2019_duplicates"
for x in lst_ids_to_remove:
    # print(x)
    loc, year = x
    name = f"swisssurface3d_{year}_{loc}.las"
    try:
        os.rename(
            os.path.join(src_originals, name),
            os.path.join(src_duplicates, name)
        )
    except:
        print("Did not work for: ", name)
```
